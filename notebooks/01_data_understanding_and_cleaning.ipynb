```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 10)
```

```python
df = pd.read_csv("../data/Telco-Customer-Churn.csv")
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

### Observation

The dataset contains **7043 customers** and **21 features**.

This provides customer demographics, subscribed services, billing information, tenure, and churn status for retention analysis.

```python
df.info()
```

### Observation

The dataset contains both numerical and categorical variables.

We'll inspect missing values and data types before analysis.

```python
df.isnull().sum()
```

```python
(df.isnull().sum() / len(df)) * 100
```

### Observation

Most columns have no missing values.

We'll also check for hidden missing values represented as blank strings.

```python
(df == " ").sum()
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df.isnull().sum()
```

### Observation

Rows with missing TotalCharges were removed because they represent a very small portion of the dataset and are unlikely to affect the analysis.

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.describe()
```

```python
df["Churn_Flag"] = df["Churn"].map({
    "Yes":1,
    "No":0
})
```

```python
df.head()
```

```python
df["Customer_Type"] = np.where(
    df["tenure"] < 12,
    "New Customer",
    "Existing Customer"
)
```

```python
df["Monthly_Spending"] = pd.cut(
    df["MonthlyCharges"],
    bins=[0,35,70,120],
    labels=["Low","Medium","High"]
)
```

```python
df.to_csv(
    "../processed_data/Telco_Customer_Churn_Cleaned.csv",
    index=False
)
```
